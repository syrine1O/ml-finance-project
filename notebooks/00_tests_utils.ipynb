```python
# La vraie variance vaut 1. On compare deux prévisions fausses.
y = np.array([1.0])
print("Sous-estime (prévoit 0.5) : MSE =", mse(y, 0.5*y), " QLIKE =", round(qlike(y, 0.5*y), 3))
print("Surestime   (prévoit 2)   : MSE =", mse(y, 2*y),   " QLIKE =", round(qlike(y, 2*y), 3))
```

```python
# Même erreur de 20 % un jour calme et un jour de crise
calme, crise = np.array([1.0]), np.array([100.0])
print("Jour calme    : MSE =", mse(calme, 1.2*calme), " QLIKE =", round(qlike(calme, 1.2*calme), 4))
print("Jour de crise : MSE =", mse(crise, 1.2*crise), " QLIKE =", round(qlike(crise, 1.2*crise), 4))
```
